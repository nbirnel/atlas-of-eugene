```python
def distribute_injuries(row):
    modes = ["PED", "PEDCYCL", "VHCL"]
    levels = ["A", "B", "C"]
    # Case 1: the mode has all of the injuries. Distribute the levels through the mode.
    for mode in modes:
        if row[f"TOT_{mode}_INJ_CNT"] == row["TOT_INJ_CNT"]:
            for level in levels:
                row[f"TOT_{mode}_INJ_LVL_{level}_CNT"] = row[
                    f"TOT_INJ_LVL_{level}_CNT"
                ]
            return row
    # Case 2: The injuries are all the same level. Distribute the injuries though the modes.
    for level in levels:
        if row[f"TOT_INJ_LVL_{level}_CNT"] == row["TOT_INJ_CNT"]:
            for mode in modes:
                row[f"TOT_{mode}_INJ_LVL_{level}_CNT"] = row[
                    f"TOT_{mode}_INJ_CNT"
                ]
            return row
    # Case 3: (There are only twelve of these in 2013-2023).
    # Distribute the most serious injuries to the most vulnerable mode.
    # This is the most realistic guess.
    # "Greedy severity-first allocation", in ChatGPT's words.
    remaining_by_level = {
        level: row[f"TOT_INJ_LVL_{level}_CNT"] for level in levels
    }
    remaining_by_mode = {mode: row[f"TOT_{mode}_INJ_CNT"] for mode in modes}

    # Allocate level A first, then B, then C
    for level in levels:
        to_assign = remaining_by_level[level]

        for mode in modes:  # PED first, then PEDCYCL, then VHCL
            if to_assign <= 0:
                break

            take = min(to_assign, remaining_by_mode[mode])
            row[f"TOT_{mode}_INJ_LVL_{level}_CNT"] += take
            remaining_by_mode[mode] -= take
            to_assign -= take

        # If something goes wrong (should not happen), leave leftovers untouched
        remaining_by_level[level] = to_assign

    return row
```

```python
def get_crashes(name, mask=None):
    cache = f"cache/odot_crashes/{name}.pickle"
    try:
        return eug.get_pickle(cache)
    except Exception as e:
        pass
    crashes = {}
    for year in range(2014, 2024):
        file = f"data/state-of-oregon/odot/{year}_Statewide_Crashes.gdb.zip"
        layers = fiona.listlayers(file)
        layer_names = [
            f"crashes{year}",
            f"Crashes_{year}",
            "crashes",
        ]
        for layer in layer_names:
            if layer in layers:
                crashes[year] = gpd.read_file(
                    file,
                    layer=layer,
                    mask=mask,
                    columns=columns,
                )
                break
        for column in crashes[year].columns:
            if column.endswith("_FLG"):
                crashes[year][column] = crashes[year][column].astype(
                    "boolean"
                )
            elif column.endswith("_DESC"):
                crashes[year][column] = crashes[year][column].astype(
                    "category"
                )
            elif column.endswith("_NM"):
                crashes[year][column] = crashes[year][column].astype(
                    "category"
                )
            elif column.endswith("_IND"):
                crashes[year][column] = crashes[year][column].astype(
                    "category"
                )

    crashes = pd.concat(crashes.values())
    crashes["CRASH_YR_NO"] = crashes["CRASH_YR_NO"].astype(int)
    crashes["TOT_VHCL_INJ_CNT"] = (
        crashes["TOT_INJ_CNT"]
        - crashes["TOT_PEDCYCL_INJ_CNT"]
        - crashes["TOT_PED_INJ_CNT"]
    )
    crashes["TOT_VHCL_FATAL_CNT"] = (
        crashes["TOT_FATAL_CNT"]
        - crashes["TOT_PEDCYCL_FATAL_CNT"]
        - crashes["TOT_PED_FATAL_CNT"]
    )
    modes = ["PED", "PEDCYCL", "VHCL"]
    for mode in modes:
        for level in ["A", "B", "C"]:
            crashes[f"TOT_{mode}_INJ_LVL_{level}_CNT"] = 0
    crashes = crashes.apply(distribute_injuries, axis=1).reset_index(
        drop=True
    )
    for mode in modes:
        crashes[f"TOT_{mode}_VICTIM_CNT"] = (
            crashes[f"TOT_{mode}_FATAL_CNT"] + crashes[f"TOT_{mode}_INJ_CNT"]
        )
    crashes["crashes"] = 1
    with open(cache, "wb") as f:
        pickle.dump(crashes, f, pickle.HIGHEST_PROTOCOL)
    return crashes
```

```python
help_html = f"""
<div style={maps.get_help_div_style()}>
Select or de-select mode types to change which crashes are displayed.<br>
Click on crashes for details.<br>

{maps.get_help_footer()}
</div>
"""
```

```python
# -----------------------------
# Marker helpers
# -----------------------------
def marker_size(row):
    return np.sqrt(
        row["TOT_FATAL_CNT"] * 64
        + row["TOT_INJ_LVL_A_CNT"] * 25
        + row["TOT_INJ_LVL_B_CNT"] * 4
        + row["TOT_INJ_LVL_C_CNT"]
    )


mode_colors = {
    "Pedestrian": "#e41a1c",  # red
    "Bike": "#377eb8",  # blue
    "Vehicle": "#4daf4a",  # green
}

mode_cols = {
    "Pedestrian": "TOT_PED_VICTIM_CNT",
    "Bike": "TOT_PEDCYCL_VICTIM_CNT",
    "Vehicle": "TOT_VHCL_VICTIM_CNT",
}
```

```python
legend_html = f"""
<div style={maps.get_legend_div_style()}>
<b>Legend</b><br>


<i style="background:{mode_colors['Vehicle']}; width:12px; height:12px; display:inline-block; margin-right:6px;"></i> Vehicle<br>
<i style="background:{mode_colors['Pedestrian']}; width:12px; height:12px; display:inline-block; margin-right:6px;"></i> Pedestrian<br>
<i style="background:{mode_colors['Bike']}; width:12px; height:12px; display:inline-block; margin-right:6px;"></i> Bike<br>
<hr style="margin:4px 0;">
Circle radius based on severity of crash and number of victims
</div>
"""
```

```python
m = folium.Map(
    location=[gdf.geometry.y.mean(), gdf.geometry.x.mean()],
    zoom_start=12,
    tiles=None,
)
# Add the tile layer explicitly so we can tell it not to show up in the layer control
folium.TileLayer("CartoDB positron", control=False).add_to(m)


# When first displayed, the least common crashes will
# sit on top of the most common crashes so they are not masked.
display_order = ["Vehicle", "Pedestrian", "Bike"]
mode_layers = {
    mode: folium.FeatureGroup(name=mode, show=True).add_to(m)
    for mode in display_order
}

## uncomment the next line for fast test run
# gdf = gdf.head(1000)

for idx, row in gdf.iterrows():
    for mode, mode_col in mode_cols.items():
        if row[mode_col] > 0:
            folium.CircleMarker(
                location=[row.geometry.y, row.geometry.x],
                radius=marker_size(row),
                color=mode_colors[mode],
                fill=True,
                fill_opacity=0.6,
                popup=folium.Popup(
                    f"Year: {row['CRASH_YR_NO']}<br>"
                    f"Mode: {mode}<br>"
                    f"Fatalities: {row['TOT_FATAL_CNT']}<br>"
                    f"Serious Injuries: {row['TOT_INJ_LVL_A_CNT']}<br>"
                    f"Medium Injuries: {row['TOT_INJ_LVL_B_CNT']}<br>"
                    f"Minor Injuries: {row['TOT_INJ_LVL_C_CNT']}<br>"
                    f"Collis. Type: {row['COLLIS_TYP_LONG_DESC']}<br>"
                    f"Intersection Type: {row['ISECT_TYP_SHORT_DESC']}<br>"
                    f"Cause: {row['CRASH_CAUSE_1_LONG_DESC']}<br>"
                    f"Event: {row['CRASH_EVNT_1_LONG_DESC']}<br>"
                    f"Crash Type: {row['CRASH_TYP_LONG_DESC']}<br>"
                    f"Posted Speed Limit: {row['POST_SPEED_LMT_VAL']}<br>"
                    f"Alcohol: {row['ALCHL_INVLV_FLG']}<br>"
                    f"Hit Run: {row['CRASH_HIT_RUN_FLG']}<br>"
                    f"Date: {row['CRASH_DT']}<br>",
                    max_width=200,
                ),
            ).add_to(mode_layers[mode])

title = f"Eugene Crashes {gdf.CRASH_YR_NO.min()} - {gdf.CRASH_YR_NO.max()}"
title_html = maps.get_title_html(title)
m.get_root().html.add_child(folium.Element(title_html))
m.get_root().html.add_child(folium.Element(help_html))
m.get_root().html.add_child(folium.Element(legend_html))
folium.LayerControl(collapsed=False).add_to(m)
Fullscreen(position="topright").add_to(m)
m.save("docs/eugene-crash-map.html")
m
```
